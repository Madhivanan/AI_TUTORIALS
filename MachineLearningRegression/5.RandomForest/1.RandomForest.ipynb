```python
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
dataset = pd.read_csv("50_Startups.csv")
```

```python
dataset = pd.get_dummies(dataset, drop_first = 1)
```

```python
x_input_dependent = dataset.drop("Profit", axis = 1)
```

```python
y_output_independent = dataset[["Profit"]]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x_input_dependent, y_output_independent, test_size = 0.30, random_state = 42)
```

```python
print(f"X Train set:\n {X_train}")
print(f"X Test set:\n {X_test}")
print(f"Y Train set:\n {y_train}")
print(f"Y Test set:\n {y_test}")
```

```python
model = RandomForestRegressor(
    n_estimators=100,
    criterion='absolute_error',
    random_state = 0
)
model.fit(X_train, y_train)
```

```python
model.get_params()
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
r2_result = r2_score(y_test, y_pred)
```

```python
r2_result
```

```python
filename = "finalized_random_forest.sav"
pickle.dump(model, open(filename, "wb"))
```
